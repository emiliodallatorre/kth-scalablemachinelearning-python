**Fit normalization on training images only.** For each channel $c$ and $M$ training images,

$$\mu_c=\frac1M\sum_i x_{ic},\qquad
\sigma_c=\sqrt{\frac1M\sum_i x_{ic}^2-\mu_c^2},\qquad
\widehat{x}_{ic}=\frac{x_{ic}-\mu_c}{\sigma_c}.$$

```python
from tqdm import tqdm

channel_sum = np.zeros(3)
channel_square_sum = np.zeros(3)
pixel_count = 0
for start in range(0, len(train_ids), 256):
    pixels = database.data[train_ids[start : start + 256]].astype(np.float64) / 255
    channel_sum += pixels.sum(axis=(0, 1, 2))
    channel_square_sum += np.square(pixels).sum(axis=(0, 1, 2))
    pixel_count += np.prod(pixels.shape[:3])

mean = channel_sum / pixel_count
std = np.sqrt(np.maximum(channel_square_sum / pixel_count - mean**2, 1e-12))
stats = {"mean": mean.tolist(), "std": std.tolist()}
display(pd.DataFrame(stats, index=["red", "green", "blue"]))
```

**Save the raw image ids.** Our small CNN uses the native 32×32 images. A dataset view returns a raw [0,1] tensor, label and original image ID.

```python
class RawImages(Dataset):
    def __init__(self, dataset, ids):
        self.dataset = dataset
        self.ids = list(map(int, ids))

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        image_id = self.ids[idx]
        image, target = self.dataset[image_id]
        return TF.to_tensor(image), int(target), image_id
```

**Prepare a batch.** Normalize using training statistics. The demo keeps `size = 32`. Part B will use an explicitly chosen input resolution for each architecture. Keeping this differentiable operation inside the model call also allows later gradients with respect to pixels.

```python
def prepare_inputs(raw, stats, size):
    if raw.shape[-2:] != (size, size):
        if raw.device.type == "mps":
            raise NotImplementedError("Antialiased resizing is not supported on this MPS version")
        raw = nn.functional.interpolate(
            raw, size=(size, size), mode="bilinear",
            align_corners=False, antialias=True,
        )
    mean = raw.new_tensor(stats["mean"])[None, :, None, None]
    std = raw.new_tensor(stats["std"])[None, :, None, None]
    return (raw - mean) / std
```

## A3 Conv2d
`nn.Conv2d` applies a 2D convolution over an input signal of size $(B, C_{in}, H_{in}, W_{in})$ and gives output of size $(B, C_{out}, H_{out}, W_{out})$, where
$$
H_{out} = \left\lfloor\frac{H_{in}+2\times \text{padding}[0] - \text{dilation}[0]\times (\text{kernel_size}[0]-1)-1}{\text{stride}[0]} + 1\right\rfloor,
$$\
$$
W_{out} = \left\lfloor\frac{W_{in}+2\times \text{padding}[1] - \text{dilation}[1]\times (\text{kernel_size}[1]-1)-1}{\text{stride}[1]} + 1\right\rfloor.
$$
From documentation:


> `class torch.nn.Conv2d(in_channels, out_channels, kernel_size, stride=1, padding=0, dilation=1, groups=1, bias=True, padding_mode='zeros', device=None, dtype=None)`

## A4 A small CNN classifier
We use a small network with two convolutional stages and a dense classification head as demonstrative example. ReLU adds nonlinearity, max pooling reduces resolution, and global average pooling produces one feature per channel. For the default shape,

$$[B,3,32,32]\to[B,16,16,16]\to[B,32,8,8]\to[B,32]\to[B,10].$$

```python
class SmallCNN(nn.Module):
    def __init__(self, n_channels=16, dropout=0.0):
        super().__init__()
        self.network = nn.Sequential(
            nn.Conv2d(3, n_channels, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(n_channels, 2 * n_channels, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
        )
        self.head = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Dropout(dropout),
            nn.Linear(2 * n_channels, 10),
        )

    def forward(self, images):
        return self.head(self.network(images))
```

**Model factory.** We can add a `build_model` function as a model factory to build model from configs more easily.

```python
def build_model(name, options=None):
    options = {} if options is None else dict(options)
    if name != "small_cnn":
        raise ValueError(name)
    return SmallCNN(**options).to(device), 32
```

**(Optional) Trace actual stage shapes.** You can add forward hooks to record module outputs.

```python
@torch.inference_mode()
def trace_shapes(model, size):
    rows, hooks = [], []

    def record(name):
        def hook(layer, inputs, output):
            rows.append({"stage": name, "shape": tuple(output.shape)})
        return hook

    for name, layer in model.named_children():
        hooks.append(layer.register_forward_hook(record(name)))
    model.eval()
    logits = model(torch.zeros(1, 3, size, size, device=device))
    for hook in hooks:
        hook.remove()
    assert logits.shape == (1, 10) and torch.isfinite(logits).all()
    return rows
```

## A5 Configure training
**Reproducible run with a seed, data order and explicit settings.** For demonstration only.

```python
import random
from dataclasses import dataclass, asdict, field
import time, json, uuid

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.use_deterministic_algorithms(True)

@dataclass
class Run:
    name: str
    model: str = "small_cnn"
    options: dict = field(default_factory=dict)
    seed: int = 42
    epochs: int = 1
    batch_size: int = 8
    lr: float = 0.01
    l2: float = 0.0
    augmentation: bool = False
```

**Dataloaders.** You may configure augmentations in Part B.

```python
def make_loaders(run, training_ids, validation_ids):
    return {
        "train": DataLoader(
            RawImages(database, training_ids),
            batch_size=run.batch_size,
            shuffle=True,
            generator=torch.Generator().manual_seed(run.seed),
        ),
        "train_eval": DataLoader(
            RawImages(database, training_ids),
            batch_size=run.batch_size
        ),
        "val": DataLoader(
            RawImages(database, validation_ids),
            batch_size=run.batch_size
        ),
    }
```

## A6 Train and validate
Cross-entropy compares raw logits with integer class labels:
$$L_{CE}=-\frac1B\sum_i\log\frac{e^{z_{i,y_i}}}{\sum_c e^{z_{i,c}}},$$
where $e_{z_i, y_i}$ is the logit of the $i$-th sample for its correct label $y_i$.

We first train with this objective alone. Regularized objectives are introduced briefly after evaluation, and you will make comparisons in Part B.

**Train a batch.** Move inputs and targets to device, clear gradients, compute the loss, backpropagate and update.

```python
def train_batch(model, batch, optimizer, run, size):
    raw, targets, _ = batch
    raw, targets = raw.to(device), targets.to(device)
    optimizer.zero_grad(set_to_none=True)
    logits = model(prepare_inputs(raw, stats, size))
    ce = nn.functional.cross_entropy(logits, targets)
    ce.backward()
    optimizer.step()
    count = len(targets)
    correct = (logits.argmax(1) == targets).sum().item()
    return ce.item() * count, ce.item() * count, correct, count
```

**Train one epoch.**

```python
def train_epoch(model, loader, optimizer, run, size, epoch):
    model.train()
    totals = np.zeros(4)
    print(f"Training {run.name}: epoch {epoch}/{run.epochs}", flush=True)
    for batch_number, batch in enumerate(loader, start=1):
        totals += train_batch(model, batch, optimizer, run, size)
        if batch_number % 200 == 0 or batch_number == len(loader):
            print(f"{run.name}: {batch_number}/{len(loader)} batches", flush=True)
    ce, objective, correct, count = totals
    return {
        "ce": ce / count,
        "objective": objective / count,
        "accuracy": correct / count,
        "steps": len(loader),
    }
```

**Define evaluation function.**

```python
@torch.inference_mode()
def evaluate(model, loader, size):
    model.eval()
    loss_sum = 0.0
    all_ids, all_targets, all_probs = [], [], []
    for raw, targets, ids in loader:
        raw, targets = raw.to(device), targets.to(device)
        logits = model(prepare_inputs(raw, stats, size))
        loss_sum += nn.functional.cross_entropy(logits, targets, reduction="sum").item()
        all_ids.extend(ids.tolist())
        all_targets.extend(targets.cpu().tolist())
        all_probs.extend(logits.softmax(1).cpu().tolist())
    probabilities = np.asarray(all_probs)
    targets = np.asarray(all_targets)
    predictions = probabilities.argmax(1)
    return {
        "ce": loss_sum / len(targets),
        "accuracy": float(np.mean(predictions == targets)),
        "ids": np.asarray(all_ids),
        "targets": targets,
        "predictions": predictions,
        "probabilities": probabilities,
    }
```

**Demonstration on a small subset.**

```python
def small_subset(ids, size):
    chosen, _ = train_test_split(
        ids, train_size=size, stratify=labels[ids], random_state=SPLIT_SEED
    )
    return np.sort(chosen)

demo_train_ids = small_subset(train_ids, 5000)
demo_val_ids = small_subset(val_ids, 500)
demo_run = Run("demonstration", model="small_cnn", epochs=10, batch_size=32,
               lr=0.02, options={"n_channels": 32})
set_seed(demo_run.seed)
demo_model, demo_size = build_model(demo_run.model, demo_run.options)
demo_loaders = make_loaders(demo_run, demo_train_ids, demo_val_ids)
demo_optimizer = torch.optim.SGD(demo_model.parameters(), lr=demo_run.lr)
demo_history = []
```

**Run the basic loop.** Each epoch updates weights and then evaluates validation images without gradients.

```python
for epoch in range(1, demo_run.epochs + 1):
    training = train_epoch(demo_model, demo_loaders["train"], demo_optimizer, demo_run, demo_size, epoch)
    validation = evaluate(demo_model, demo_loaders["val"], demo_size)
    demo_history.append({
        "epoch": epoch,
        "train_ce": training["ce"],
        "train_accuracy": training["accuracy"],
        "val_ce": validation["ce"],
        "val_accuracy": validation["accuracy"],
    })
display(pd.DataFrame(demo_history))
```

**Capacity and regularization.** For CNNs, capacity is influenced by:


*   Depth. Number of convolutional layers.
*   Width. Number of channels (width) per layer.
*   Kernel size. Larger kernels can capture broader spatial patterns.
*   Architecture. Residual connections, attention mechanisms, etc.

Regularization can reduce overfitting by constraining or modifying how the model learns. Common techniques include L2 loss (penalizes large weights), dropout (randomly disables activations during training) and data augmentation (introduces variations of training images). The L2 regularization is given by
$$L=L_{CE}+\frac{\lambda}{2}\sum_W\| W\|_F^2,$$
where $L_{CE}$ is the cross-entropy loss, $W$ is the weight matrix, $\|\cdot\|_F^2$ denotes the squared Forbenius norm, and $\lambda$ is the regularization strength. For a weight matrix,
$$
\|W\|_F^2 = \sum{i,j}W_{ij}^2.
$$
For convolutional kernels, the sum extends over all kernel dimensions. It is typically applied to the trainable weights of convolutional and fully connected layers.

**One pixel-gradient adversarial.** A one-step Fast Gradient Sign Method (FGSM) perturbation increases the labelled loss within a raw-pixel bound.
$$x_{adv}=\operatorname{clip}_{[0,1]}(x+\epsilon\operatorname{sign}(\nabla_x L))$$
where:
- $x$: Original image, with pixel values in [0,1].
- $L$: Classification loss, typically cross-entropy.
- $\nabla_x L$: Gradient of loss with respect to the input pixels, not model weights.
- $\operatorname{sign}$: Converts each gradient element to -1, 0, or +1.
- $\epsilon$: Maximum permitted change to each pixel.
- $\operatorname{clip}_{[0,1]}$: Ensures resulting pixel values remain valid.

```python
demo_model.eval()
attack_input = raw_demo.unsqueeze(0).to(device).requires_grad_(True)
attack_target = torch.tensor([target_demo], device=device)
attack_logits = demo_model(prepare_inputs(attack_input, stats, demo_size))
attack_loss = nn.functional.cross_entropy(attack_logits, attack_target)
(pixel_gradient,) = torch.autograd.grad(attack_loss, attack_input)
epsilon_demo = 4 / 255
perturbed_demo = (attack_input + epsilon_demo * pixel_gradient.sign()).clamp(0, 1).detach()
print("Maximum pixel change:", (perturbed_demo - attack_input.detach()).abs().max().item())
print("Clean and perturbed predictions:")
print(predict(demo_model, torch.cat([attack_input.detach(), perturbed_demo]), demo_size).argmax(1))
```

### AlexNet

AlexNet was introduced by Krizhevsky et al. (2012). It consists of five convolutional layers followed by three fully connected layers. The original architecture uses ReLU activation functions, max pooling, and dropout.

Reference: [ImageNet Classification with Deep Convolutional Neural Networks](https://proceedings.neurips.cc/paper/2012/hash/c399862d3b9d6b76c8436e924a68c45b-Abstract.html)

#### Architecture (adapted)

AlexNet follows a sequential architecture:

**Input → Conv1 → Pool → Conv2 → Pool → Conv3 → Conv4 → Conv5 → Pool → Flatten → FC1 → FC2 → FC3**

The original AlexNet was designed for approximately $224 \times 224$ RGB images. Its first convolution uses an $11 \times 11$ kernel with stride 4, which reduces the spatial dimensions substantially.

For CIFAR-10, the input images are only $32 \times 32$. Using the original configuration would reduce the spatial dimensions too quickly. We therefore use smaller kernels and adjust the pooling operations.

#### Suggested CIFAR-10 adaptation

| Layer | Configuration | Output shape (C × H × W) |
|---|---|---|
| Input | RGB image | 3 × 32 × 32 |
| Conv1 | 64 filters, 3×3, stride 1, padding 1 + ReLU | 64 × 32 × 32 |
| MaxPool1 | 2×2, stride 2 | 64 × 16 × 16 |
| Conv2 | 192 filters, 3×3, padding 1 + ReLU | 192 × 16 × 16 |
| MaxPool2 | 2×2, stride 2 | 192 × 8 × 8 |
| Conv3 | 384 filters, 3×3, padding 1 + ReLU | 384 × 8 × 8 |
| Conv4 | 256 filters, 3×3, padding 1 + ReLU | 256 × 8 × 8 |
| Conv5 | 256 filters, 3×3, padding 1 + ReLU | 256 × 8 × 8 |
| MaxPool3 | 2×2, stride 2 | 256 × 4 × 4 |
| Flatten | Flatten feature maps | 4096 |
| FC1 | 4096 → 1024 + ReLU + Dropout | 1024 |
| FC2 | 1024 → 512 + ReLU + Dropout | 512 |
| FC3 | 512 → 10 | 10 |

You may use a dropout probability of 0.5 in the fully connected layers, as in the original AlexNet.

#### Implementation considerations

- Apply ReLU after each convolutional layer and the first two fully connected layers.
- Use max pooling to progressively reduce spatial resolution.
- Flatten the final convolutional feature maps before passing them to the fully connected layers.
- The final layer should output 10 raw logits, one for each CIFAR-10 class. Do not apply softmax before `nn.CrossEntropyLoss`.
- You may experiment with changing the number of filters or the size of the fully connected layers to investigate model capacity.

### ResNet

ResNet (Residual Network) was introduced by He et al. (2015). Its main innovation is the **residual connection**, also called a skip connection, which makes deeper networks easier to optimize.

Reference: [Deep Residual Learning for Image Recognition](https://arxiv.org/abs/1512.03385)

#### Residual learning

In a conventional CNN, a sequence of layers learns a transformation $H(x)$ directly.

In ResNet, the layers learn a residual transformation $F(x)$, and the original input is added back:

$$H(x)=F(x)+x$$

This creates a direct path through which information can propagate across layers.

For a basic residual block, the transformation is approximately:

**Input → Conv → BatchNorm → ReLU → Conv → BatchNorm → Add input → ReLU**

If the input and output dimensions are different, the shortcut connection must transform the input to match the output dimensions. A common solution is a $1 \times 1$ convolution, optionally with stride 2, followed by batch normalization.

#### ResNet-18 architecture (adpated)

ResNet-18 has 18 learnable layers: 17 convolutional layers and one fully connected layer. Its convolutional layers are organized into four stages, with two residual blocks per stage.

Each basic residual block contains two $3 \times 3$ convolutions.

For CIFAR-10, here is a recommended adaptation:

| Stage | Configuration | Output shape (C × H × W) |
|---|---|---|
| Input | RGB image | 3 × 32 × 32 |
| Initial Conv | 3×3, 64 filters, stride 1, padding 1 + BN + ReLU | 64 × 32 × 32 |
| Stage 1 | 2 residual blocks, 64 channels | 64 × 32 × 32 |
| Stage 2 | 2 residual blocks, 128 channels | 128 × 16 × 16 |
| Stage 3 | 2 residual blocks, 256 channels | 256 × 8 × 8 |
| Stage 4 | 2 residual blocks, 512 channels | 512 × 4 × 4 |
| Global Average Pooling | Average each feature map spatially | 512 × 1 × 1 |
| Flatten | Flatten pooled features | 512 |
| FC | 512 → 10 | 10 |

Unlike the original ImageNet ResNet-18, this adaptation uses a $3 \times 3$ initial convolution with stride 1 and removes the initial max-pooling layer. This preserves spatial information for the smaller CIFAR-10 images.

#### Basic residual block

Implement a reusable `BasicBlock` module containing:

1. A $3 \times 3$ convolution followed by batch normalization and ReLU.
2. A second $3 \times 3$ convolution followed by batch normalization.
3. A shortcut connection that adds the input to the output of the second batch normalization.
4. A final ReLU activation after the addition.

When the spatial dimensions and channel counts remain unchanged, use an identity shortcut:

$$y=\operatorname{ReLU}(F(x)+x)$$

When the dimensions differ, use a projection shortcut $W_s$:

$$y=\operatorname{ReLU}(F(x)+W_s(x))$$

For this implementation, use a $1 \times 1$ convolution and batch normalization for the projection shortcut.

#### Constructing the four stages

- **Stage 1:** Two residual blocks with 64 output channels. Both blocks use stride 1.
- **Stage 2:** Two residual blocks with 128 output channels. The first block uses stride 2 in its first convolution and shortcut to downsample; the second uses stride 1.
- **Stage 3:** Two residual blocks with 256 output channels, following the same downsampling pattern.
- **Stage 4:** Two residual blocks with 512 output channels, following the same downsampling pattern.

Finally, apply global average pooling and a fully connected classification layer.

#### Implementation considerations

- Implement the residual block as a separate PyTorch module and reuse it throughout the network.
- Ensure that the tensors being added in each residual connection have identical shapes.
- Use batch normalization after each convolution in the residual branches.
- Do not apply ReLU immediately before the residual addition; the second convolution is followed by batch normalization, then addition, then ReLU.
- Use global average pooling instead of large fully connected layers.

Implement and train AlexNet and ResNet using PyTorch layers, starting from the architectures and CIFAR-10 adaptations discussed in A9.
```python
# Task 2: implement the A9 architectures directly from PyTorch layers.
class CIFARAlexNet(nn.Module):
    def __init__(self, fc1=1024, dropout=0.5):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(64, 192, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(192, 384, 3, padding=1), nn.ReLU(),
            nn.Conv2d(384, 256, 3, padding=1), nn.ReLU(),
            nn.Conv2d(256, 256, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(), nn.Linear(256 * 4 * 4, fc1), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(fc1, 512), nn.ReLU(), nn.Dropout(dropout), nn.Linear(512, 10),
        )

    def forward(self, x):
        return self.classifier(self.features(x))

class BasicBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()
        self.branch = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, stride=stride, padding=1, bias=False),
            nn.BatchNorm2d(out_channels), nn.ReLU(),
            nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
        )
        self.shortcut = (
            nn.Identity() if stride == 1 and in_channels == out_channels
            else nn.Sequential(
                nn.Conv2d(in_channels, out_channels, 1, stride=stride, bias=False),
                nn.BatchNorm2d(out_channels),
            )
        )
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.relu(self.branch(x) + self.shortcut(x))

class CIFARResNet18(nn.Module):
    def __init__(self):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1, bias=False),
            nn.BatchNorm2d(64), nn.ReLU(),
        )
        self.stage1 = nn.Sequential(BasicBlock(64, 64), BasicBlock(64, 64))
        self.stage2 = nn.Sequential(BasicBlock(64, 128, 2), BasicBlock(128, 128))
        self.stage3 = nn.Sequential(BasicBlock(128, 256, 2), BasicBlock(256, 256))
        self.stage4 = nn.Sequential(BasicBlock(256, 512, 2), BasicBlock(512, 512))
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Linear(512, 10)
        for layer in self.modules():
            if isinstance(layer, nn.Conv2d):
                nn.init.kaiming_normal_(layer.weight, mode="fan_out", nonlinearity="relu")
            elif isinstance(layer, nn.BatchNorm2d):
                nn.init.ones_(layer.weight)
                nn.init.zeros_(layer.bias)

    def forward(self, x):
        x = self.stem(x)
        x = self.stage1(x)
        x = self.stage2(x)
        x = self.stage3(x)
        x = self.stage4(x)
        return self.fc(torch.flatten(self.pool(x), 1))

def build_model(name, options=None):
    options = {} if options is None else dict(options)
    constructors = {"small_cnn": SmallCNN, "alexnet": CIFARAlexNet,
                    "resnet18": CIFARResNet18}
    if name not in constructors:
        raise ValueError(f"Unknown model: {name}")
    return constructors[name](**options).to(device), 32

class AugmentedRawImages(RawImages):
    def __init__(self, dataset, ids):
        super().__init__(dataset, ids)
        self.augment = transforms.Compose([
            transforms.RandomCrop(32, padding=4),
            transforms.RandomHorizontalFlip(),
        ])

    def __getitem__(self, idx):
        raw, target, image_id = super().__getitem__(idx)
        return self.augment(raw), target, image_id

def train_configuration(name, model_name, options=None, *, l2=0.0, augment=False):
    run = Run(name=name, model=model_name, options=options or {}, seed=42, epochs=1,
              batch_size=32, lr=0.01, l2=l2, augmentation=augment)
    set_seed(run.seed)
    model, size = build_model(run.model, run.options)
    loaders = make_loaders(run, train_ids, val_ids)
    if augment:
        loaders["train"] = DataLoader(
            AugmentedRawImages(database, train_ids), batch_size=run.batch_size,
            shuffle=True, generator=torch.Generator().manual_seed(run.seed),
        )
    optimizer = torch.optim.SGD(model.parameters(), lr=run.lr, momentum=0.9,
                                weight_decay=run.l2)
    params = sum(parameter.numel() for parameter in model.parameters())
    initial_train = evaluate(model, loaders["train_eval"], size)
    initial_val = evaluate(model, loaders["val"], size)
    history = [{
        "epoch": 0, "train_ce": initial_train["ce"],
        "train_accuracy": initial_train["accuracy"],
        "val_ce": initial_val["ce"], "val_accuracy": initial_val["accuracy"],
    }]
    best_state, best_accuracy, best_epoch = None, float("-inf"), None
    training_seconds = 0.0
    for epoch in range(1, run.epochs + 1):
        synchronize()
        started = time.perf_counter()
        train_epoch(model, loaders["train"], optimizer, run, size, epoch)
        synchronize()
        training_seconds += time.perf_counter() - started
        training = evaluate(model, loaders["train_eval"], size)
        validation = evaluate(model, loaders["val"], size)
        history.append({
            "epoch": epoch, "train_ce": training["ce"],
            "train_accuracy": training["accuracy"],
            "val_ce": validation["ce"], "val_accuracy": validation["accuracy"],
        })
        if validation["accuracy"] > best_accuracy:
            best_accuracy, best_epoch = validation["accuracy"], epoch
            best_state = {key: value.detach().cpu().clone()
                          for key, value in model.state_dict().items()}
    result = {
        "name": name, "model": model_name, "options": dict(run.options),
        "l2": l2, "augmentation": augment, "parameters": params,
        "train_images": len(train_ids), "training_seconds": training_seconds,
        "best_epoch": best_epoch, "validation_accuracy": best_accuracy,
        "history": history, "state": best_state,
    }
    del model, optimizer, loaders
    if device.type == "mps":
        torch.mps.empty_cache()
    return result

def restore_model(result):
    model, size = build_model(result["model"], result["options"])
    model.load_state_dict(result["state"])
    return model, size

def summary_table(results):
    return pd.DataFrame([{
        "configuration": r["name"], "parameters": r["parameters"],
        "training images": r["train_images"], "training seconds": r["training_seconds"],
        "best epoch": r["best_epoch"], "validation accuracy": r["validation_accuracy"],
    } for r in results])

print("Full-data budget: 45,000 training images and one epoch per configuration.")
print("Model selection: highest validation accuracy among completed epochs; state kept in memory.")
baselines = [
    train_configuration("AlexNet", "alexnet"),
    train_configuration("ResNet-18", "resnet18"),
]
display(summary_table(baselines))
for result in baselines:
    model, size = restore_model(result)
    print(result["name"], "feature shapes:")
    display(pd.DataFrame(trace_shapes(model, size)))
    del model
display(Markdown(
    "Both models use raw logits, cross-entropy, SGD (learning rate 0.01, momentum 0.9), "
    "the same split, seed, batch size and one-epoch image budget. The AlexNet stem "
    "uses 3 x 3 kernels and 2 x 2 pooling, unlike the large-stride ImageNet stem; "
    "its dense head is smaller. ResNet replaces the ImageNet stem and max pool "
    "with a 3 x 3 stride-1 stem. Its residual shortcuts add the input or a "
    "projected input to each residual branch, aiding gradient flow. One epoch and "
    "one seed cannot establish consistent superiority. Peak GPU memory is not "
    "measured by this experiment."
))
```

```python
import copy
import io
import statistics
import time
import warnings

import torch.ao.quantization as quantization

# Select a supported CPU quantization backend.
supported_engines = torch.backends.quantized.supported_engines
quant_engine = (
    "x86" if "x86" in supported_engines
    else "fbgemm" if "fbgemm" in supported_engines
    else "qnnpack" if "qnnpack" in supported_engines
    else None
)
if quant_engine is None:
    raise RuntimeError(
        f"No supported quantization backend found: {supported_engines}"
    )

# Required because this PyTorch quantized CPU operator
# does not implement deterministic execution.
torch.use_deterministic_algorithms(False)

torch.backends.quantized.engine = quant_engine
print("Quantization backend:", quant_engine)


class QuantizableAlexNet(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.quant = quantization.QuantStub()
        self.model = model
        self.dequant = quantization.DeQuantStub()

    def forward(self, inputs):
        inputs = self.quant(inputs)
        outputs = self.model(inputs)
        return self.dequant(outputs)


# Restore the trained AlexNet checkpoint from Task 2.
assert "baselines" in globals() and len(baselines) >= 2
assert baselines[0]["model"] == "alexnet"

alexnet_fp32, alexnet_size = restore_model(baselines[0])
alexnet_fp32 = alexnet_fp32.cpu().eval()

# Construct and fuse the quantization-compatible model.
quantizable_alexnet = QuantizableAlexNet(
    copy.deepcopy(alexnet_fp32)
).cpu().eval()

quantization.fuse_modules(
    quantizable_alexnet.model.features,
    [
        ["0", "1"],
        ["3", "4"],
        ["6", "7"],
        ["8", "9"],
        ["10", "11"],
    ],
    inplace=True,
)
quantization.fuse_modules(
    quantizable_alexnet.model.classifier,
    [
        ["1", "2"],
        ["4", "5"],
    ],
    inplace=True,
)

quantizable_alexnet.qconfig = quantization.get_default_qconfig(
    quant_engine
)

prepared_alexnet = quantization.prepare(
    quantizable_alexnet,
    inplace=False,
).cpu().eval()

calibration_loader = DataLoader(
    RawImages(database, train_ids),
    batch_size=32,
    shuffle=False,
    num_workers=0,
)

calibration_started = time.perf_counter()
with torch.inference_mode():
    for raw_images, _, _ in calibration_loader:
        inputs = prepare_inputs(raw_images, stats, alexnet_size)
        prepared_alexnet(inputs)
calibration_seconds = time.perf_counter() - calibration_started
print(f"Calibration time: {calibration_seconds:.2f} seconds")

conversion_started = time.perf_counter()
quantized_alexnet = quantization.convert(
    prepared_alexnet,
    inplace=False,
).cpu().eval()
conversion_seconds = time.perf_counter() - conversion_started
print(f"Conversion time: {conversion_seconds:.2f} seconds")
print("Static quantization completed.")


def challenge1_evaluate_cpu(model, loader, size):
    model.eval()
    all_targets = []
    all_predictions = []
    loss_sum = 0.0

    with torch.inference_mode():
        for raw_images, targets, _ in loader:
            inputs = prepare_inputs(raw_images, stats, size)
            logits = model(inputs)
            loss_sum += nn.functional.cross_entropy(
                logits.float(), targets, reduction="sum"
            ).item()
            all_targets.extend(targets.tolist())
            all_predictions.extend(logits.argmax(dim=1).tolist())

    targets_array = np.asarray(all_targets)
    predictions_array = np.asarray(all_predictions)
    return {
        "accuracy": float(np.mean(predictions_array == targets_array)),
        "cross_entropy": loss_sum / len(targets_array),
    }


challenge1_validation_loader = DataLoader(
    RawImages(database, val_ids),
    batch_size=32,
    shuffle=False,
    num_workers=0,
)

# Evaluate both models on the same validation images, on the CPU.
fp32_validation = challenge1_evaluate_cpu(
    alexnet_fp32, challenge1_validation_loader, alexnet_size
)

int8_validation = challenge1_evaluate_cpu(
    quantized_alexnet, challenge1_validation_loader, alexnet_size
)


def serialized_state_dict_size_mb(model):
    buffer = io.BytesIO()
    torch.save(model.state_dict(), buffer)
    return len(buffer.getvalue()) / (1024 ** 2)


def benchmark_cpu_latency_ms(model, inputs, warmup=5, repeats=30):
    model.eval()
    with torch.inference_mode():
        for _ in range(warmup):
            model(inputs)

        measurements = []
        for _ in range(repeats):
            started = time.perf_counter()
            model(inputs)
            measurements.append((time.perf_counter() - started) * 1000)

    return {
        "mean": float(np.mean(measurements)),
        "median": float(np.median(measurements)),
    }


timing_raw, _, _ = next(iter(challenge1_validation_loader))
timing_inputs = prepare_inputs(timing_raw, stats, alexnet_size)

fp32_latency = benchmark_cpu_latency_ms(alexnet_fp32, timing_inputs)
int8_latency = benchmark_cpu_latency_ms(quantized_alexnet, timing_inputs)

fp32_size_mb = serialized_state_dict_size_mb(alexnet_fp32)
int8_size_mb = serialized_state_dict_size_mb(quantized_alexnet)

challenge1_results = pd.DataFrame([
    {
        "Model": "AlexNet FP32",
        "Precision": "FP32",
        "Validation accuracy": fp32_validation["accuracy"],
        "Serialized size (MB)": fp32_size_mb,
        "Mean CPU latency (ms / batch)": fp32_latency["mean"],
        "Median CPU latency (ms / batch)": fp32_latency["median"],
    },
    {
        "Model": "AlexNet INT8",
        "Precision": "Static INT8",
        "Validation accuracy": int8_validation["accuracy"],
        "Serialized size (MB)": int8_size_mb,
        "Mean CPU latency (ms / batch)": int8_latency["mean"],
        "Median CPU latency (ms / batch)": int8_latency["median"],
    },
])

storage_reduction = (
    (fp32_size_mb - int8_size_mb) / fp32_size_mb * 100
)
accuracy_loss_pp = (
    fp32_validation["accuracy"] - int8_validation["accuracy"]
) * 100
latency_speedup = fp32_latency["mean"] / int8_latency["mean"]

display(challenge1_results)
print(f"Storage reduction: {storage_reduction:.2f}%")
print(f"Accuracy loss: {accuracy_loss_pp:.2f} percentage points")
print(f"Latency speedup: {latency_speedup:.2f}x")
print(f"Calibration time: {calibration_seconds:.2f} seconds")
print(f"Conversion time: {conversion_seconds:.2f} seconds")

# Restore deterministic execution
torch.use_deterministic_algorithms(True)
```

The following cells evaluate the Task 2 ResNet-18 checkpoint on the unchanged validation set, then on Gaussian-blurred versions of the same images.

```python
assert "baselines" in globals(), "Run Task 2 before Challenge 2."
assert len(baselines) >= 2, "Both Task 2 baselines must be available."
assert baselines[1]["model"] == "resnet18", (
    "baselines[1] must contain the ResNet-18 baseline."
)

challenge2_model, CHALLENGE2_INPUT_SIZE = restore_model(baselines[1])
challenge2_model.eval()

CHALLENGE2_BATCH_SIZE = 128

challenge2_val_loader = DataLoader(
    RawImages(database, val_ids),
    batch_size=CHALLENGE2_BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

print("Challenge 2 model: Task 2 ResNet-18")
print("Device:", device)
print("Input size:", CHALLENGE2_INPUT_SIZE)
print("Validation images:", len(val_ids))
print(
    "Task 2 recorded validation accuracy:",
    f"{baselines[1]['validation_accuracy']:.2%}",
)
```

```python
@torch.inference_mode()
def challenge2_evaluate(loader, sigma=None):
    challenge2_model.eval()

    all_ids = []
    all_targets = []
    all_predictions = []
    loss_sum = 0.0

    for raw_images, targets, image_ids in loader:
        raw_images = raw_images.to(device)
        targets = targets.to(device)

        if sigma is not None:
            raw_images = TF.gaussian_blur(
                raw_images,
                kernel_size=[5, 5],
                sigma=[float(sigma), float(sigma)],
            )

        inputs = prepare_inputs(
            raw_images,
            stats,
            CHALLENGE2_INPUT_SIZE,
        )
        logits = challenge2_model(inputs)

        loss_sum += nn.functional.cross_entropy(
            logits, targets, reduction="sum"
        ).item()

        predictions = logits.argmax(dim=1)

        all_ids.extend(image_ids.tolist())
        all_targets.extend(targets.cpu().tolist())
        all_predictions.extend(predictions.cpu().tolist())

    ids = np.asarray(all_ids)
    targets = np.asarray(all_targets)
    predictions = np.asarray(all_predictions)

    return {
        "ids": ids,
        "targets": targets,
        "predictions": predictions,
        "loss": loss_sum / len(targets),
        "accuracy": float(np.mean(predictions == targets)),
    }


challenge2_clean = challenge2_evaluate(
    challenge2_val_loader,
    sigma=None,
)

challenge2_sigmas = [0.5, 1.0, 1.5]
challenge2_outputs = {"clean": challenge2_clean}
challenge2_rows = [{
    "condition": "Clean",
    "accuracy (%)": 100 * challenge2_clean["accuracy"],
    "accuracy drop (pp)": 0.0,
    "accuracy retained (%)": 100.0,
    "changed predictions": 0,
    "cross-entropy": challenge2_clean["loss"],
}]

for sigma in challenge2_sigmas:
    result = challenge2_evaluate(
        challenge2_val_loader,
        sigma=sigma,
    )

    assert np.array_equal(challenge2_clean["ids"], result["ids"])
    assert np.array_equal(
        challenge2_clean["targets"], result["targets"]
    )

    challenge2_outputs[f"sigma={sigma}"] = result

    accuracy_drop = (
        challenge2_clean["accuracy"] - result["accuracy"]
    ) * 100

    accuracy_retained = (
        100 * result["accuracy"] / challenge2_clean["accuracy"]
        if challenge2_clean["accuracy"] > 0
        else np.nan
    )

    changed_predictions = int(np.count_nonzero(
        challenge2_clean["predictions"] != result["predictions"]
    ))

    challenge2_rows.append({
        "condition": f"Gaussian blur sigma={sigma}",
        "accuracy (%)": 100 * result["accuracy"],
        "accuracy drop (pp)": accuracy_drop,
        "accuracy retained (%)": accuracy_retained,
        "changed predictions": changed_predictions,
        "cross-entropy": result["loss"],
    })

challenge2_table = pd.DataFrame(challenge2_rows)

print(
    "Clean validation accuracy:",
    f"{challenge2_clean['accuracy']:.2%}",
)
print(
    "Task 2 recorded validation accuracy:",
    f"{baselines[1]['validation_accuracy']:.2%}",
)
display(challenge2_table)
```

```python
# Display validation images correctly classified when clean
# but incorrectly classified after Gaussian blur.

def show_challenge2_examples(sigma=1.5, max_examples=8):
    key = f"sigma={sigma}"

    if key not in challenge2_outputs:
        raise ValueError(
            f"No evaluation results found for sigma={sigma}. "
            "Run the evaluation cell first."
        )

    blurred_result = challenge2_outputs[key]

    clean_correct = (
        challenge2_clean["predictions"]
        == challenge2_clean["targets"]
    )

    blurred_wrong = (
        blurred_result["predictions"]
        != blurred_result["targets"]
    )

    positions = np.flatnonzero(
        clean_correct & blurred_wrong
    )[:max_examples]

    if len(positions) == 0:
        print(
            f"No clean-correct / blurred-wrong examples "
            f"found at sigma={sigma}."
        )
        return

    ncols = 4
    nrows = 2
    fig, axes = plt.subplots(
        nrows,
        ncols * 2,
        figsize=(16, 7),
        squeeze=False,
    )

    for i, position in enumerate(positions):
        row = i // ncols
        col = (i % ncols) * 2

        image_id = int(challenge2_clean["ids"][position])

        raw_image, true_label, _ = RawImages(
            database, [image_id]
        )[0]

        blurred_image = TF.gaussian_blur(
            raw_image,
            kernel_size=[5, 5],
            sigma=[float(sigma), float(sigma)],
        )

        clean_prediction = int(
            challenge2_clean["predictions"][position]
        )
        blurred_prediction = int(
            blurred_result["predictions"][position]
        )

        axes[row, col].imshow(
            raw_image.permute(1, 2, 0).numpy()
        )
        axes[row, col].set_title(
            f"Clean\nTrue: {class_names[true_label]}\n"
            f"Pred: {class_names[clean_prediction]}",
            fontsize=9,
        )

        axes[row, col + 1].imshow(
            blurred_image.permute(1, 2, 0).numpy()
        )
        axes[row, col + 1].set_title(
            f"Blurred ($\\sigma={sigma}$)\n"
            f"True: {class_names[true_label]}\n"
            f"Pred: {class_names[blurred_prediction]}",
            fontsize=9,
        )

    for ax in axes.flat:
        ax.axis("off")

    fig.suptitle(
        "ResNet-18 predictions on clean and blurred images",
        fontsize=14,
    )
    fig.tight_layout()
    plt.show()


show_challenge2_examples(sigma=1.5, max_examples=8)
```
